### Bai1
Yêu cầu 
Xây dựng lớp Shape với phương thức area(). 
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area(). 
Lấy π = 3.14. 
 
Input 
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE) 
Dòng 2: tham số tương ứng. 
 
Output 
In ra diện tích với 2 chữ số thập phân.

```python
import sys
class Shape:
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
        
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
        
    def area(self):
        return 3.14 * (self.r ** 2)

def main():

    input_data = sys.stdin.read().splitlines()
    if not input_data:
        return
    

    shape_type = input_data[0].strip().upper()
    

    if shape_type == "RECTANGLE":

        params = list(map(float, input_data[1].split()))
        w, h = params[0], params[1]
        shape = Rectangle(w, h)
        
    elif shape_type == "CIRCLE":

        r = float(input_data[1].strip())
        shape = Circle(r)
    else:
        return
    print(f"{shape.area():.2f}")

if __name__ == "__main__":
    main()
```

Yêu cầu 
Xây dựng lớp Vehicle(brand, speed). 
Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle. 
Viết phương thức info() cho mỗi lớp: - Vehicle: brand-speed - Car: brand-speed-seats 
 
Input 
Dòng 1: loại đối tượng (Vehicle hoặc Car) 
Dòng 2: dữ liệu tương ứng. 
 
Output 
In ra chuỗi info(). 

```python
import sys
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def info(self):
        return f"{self.brand}-{self.speed}"

# Định nghĩa lớp Car kế thừa từ Vehicle
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        # Gọi hàm khởi tạo của lớp cha (Vehicle)
        super().__init__(brand, speed)
        self.seats = seats

    # Ghi đè phương thức info()
    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"

def main():
    # Đọc dữ liệu từ đầu vào chuẩn
    input_data = sys.stdin.read().splitlines()
    if not input_data:
        return
    
    # Lấy loại đối tượng (VEHICLE hoặc CAR)
    obj_type = input_data[0].strip().upper()
    
    # Tách các tham số từ dòng dữ liệu thứ 2
    params = input_data[1].split()
    
    if obj_type == "VEHICLE":
        brand = params[0]
        speed = params[1]  # Giữ nguyên định dạng chuỗi theo đề bài (brand-speed)
        obj = Vehicle(brand, speed)
        
    elif obj_type == "CAR":
        brand = params[0]
        speed = params[1]
        seats = params[2]
        obj = Car(brand, speed, seats)
        
    else:
        return

    # In ra chuỗi thông tin thông qua phương thức info()
    print(obj.info())

if __name__ == "__main__":
    main()
```

Xây dựng lớp Polygon với phương thức perimeter(). 
Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c. 
 
Input 
Một dòng chứa 3 số nguyên a b c. 
 
Output 
In ra chu vi tam giác.

```python
import sys

# Định nghĩa lớp cơ sở Polygon
class Polygon:
    def perimeter(self):
        pass

# Định nghĩa lớp Triangle kế thừa từ Polygon
class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
        
    # Ghi đè phương thức perimeter() của lớp cha
    def perimeter(self):
        return self.a + self.b + self.c

def main():
    # Đọc dữ liệu từ đầu vào chuẩn
    input_line = sys.stdin.read().strip()
    if not input_line:
        return
    
    # Tách chuỗi thành 3 số nguyên a, b, c
    a, b, c = map(int, input_line.split())
    
    # Khởi tạo đối tượng Triangle
    triangle = Triangle(a, b, c)
    
    # In ra chu vi tam giác
    print(triangle.perimeter())

if __name__ == "__main__":
    main()
```

### Bai 5
Yêu cầu 
Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance(). 
Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức 
add_interest() để cộng balance*rate vào số dư. 
 
Input 
balance rate 
 
Output 
Số dư sau khi cộng lãi, in 2 chữ số thập phân.  

```python
import sys

# Định nghĩa lớp cơ sở Account
class Account:
    def __init__(self, balance):
        self.balance = float(balance)

    def deposit(self, x):
        self.balance += float(x)

    def get_balance(self):
        return self.balance

# Định nghĩa lớp SavingAccount kế thừa từ Account
class SavingAccount(Account):
    def __init__(self, balance, rate):
        # Gọi hàm khởi tạo của lớp cha để gán thuộc tính balance
        super().__init__(balance)
        self.rate = float(rate)

    # Thêm phương thức tính và cộng lãi vào số dư
    def add_interest(self):
        interest = self.balance * self.rate
        self.deposit(interest) # Tái sử dụng phương thức deposit của lớp cha

def main():
    # Đọc dữ liệu từ đầu vào chuẩn
    input_line = sys.stdin.read().strip()
    if not input_line:
        return
    
    # Tách dữ liệu thành số dư ban đầu và tỷ lệ lãi suất
    balance, rate = map(float, input_line.split())
    
    # Khởi tạo đối tượng tài khoản tiết kiệm
    saving_acc = SavingAccount(balance, rate)
    
    # Thực hiện cộng lãi
    saving_acc.add_interest()
    
    # In ra số dư cuối cùng với 2 chữ số thập phân
    print(f"{saving_acc.get_balance():.2f}")

if __name__ == "__main__":
    main()
```

Xây dựng lớp Printer với phương thức print_value(x). 
Xây dựng lớp BinaryPrinter kế thừa từ Printer, ghi đè print_value(x) để in biểu diễn nhị phân của 
số nguyên dương x. 
 
Input 
Một số nguyên dương n. 
 
Output 
In ra dạng nhị phân của n.

```python
import sys

# Định nghĩa lớp cơ sở Printer
class Printer:
    def print_value(self, x):
        print(x)

# Định nghĩa lớp BinaryPrinter kế thừa từ Printer
class BinaryPrinter(Printer):
    # Ghi đè phương thức print_value(x) để đổi sang hệ nhị phân
    def print_value(self, x):
        # bin(x) trả về chuỗi có dạng '0b1010', ta dùng [2:] để cắt bỏ '0b'
        binary_string = bin(x)[2:]
        print(binary_string)

def main():
    # Đọc dữ liệu đầu vào từ bàn phím / file test
    input_data = sys.stdin.read().strip()
    if not input_data:
        return
    
    # Chuyển đổi dữ liệu nhập vào thành số nguyên dương n
    n = int(input_data)
    
    # Khởi tạo đối tượng của lớp con BinaryPrinter
    printer = BinaryPrinter()
    
    # Gọi phương thức đã được ghi đè để in ra dạng nhị phân
    printer.print_value(n)

if __name__ == "__main__":
    main()
```

### Bai 7
Yêu cầu 
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận: 
[a b] 
[c d] 
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc. 
 
Input 
Một dòng chứa 4 số nguyên a b c d. 
 
Output 
In ra trace và determinant trên cùng một dòng.

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def trace(self):
        return self.a + self.d

    def determinant(self):
        return (self.a * self.d) - (self.b * self.c)

# --- Xử lý Input và Output ---
if __name__ == "__main__":
    # Đọc dữ liệu vào từ một dòng, tách thành 4 số nguyên
    a, b, c, d = map(int, input().split())
    
    # Khởi tạo đối tượng ma trận
    matrix = Matrix2x2(a, b, c, d)
    
    # In ra kết quả trên cùng một dòng
    print(matrix.trace(), matrix.determinant())
```

#### Bai 10
Yêu cầu 
Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất: 
P(x) = ax + b, Q(x) = cx + d. 
Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m. 
 
Input 
a b c d 
 
Output 
Chuỗi của đa thức P(x)+Q(x).

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def sum_str(self):
        # Tính hệ số của đa thức tổng: (a + c)x + (b + d)
        k = self.a + self.c
        m = self.b + self.d
        
        # Định dạng chuỗi hiển thị theo mẫu kx + m
        # Lưu ý: Sử dụng dấu + hoặc - tự động dựa vào giá trị của m
        if m >= 0:
            return f"{k}x + {m}"
        else:
            return f"{k}x - {abs(m)}"

# --- Xử lý Input và Output ---
if __name__ == "__main__":
    # Đọc 4 số a, b, c, d từ một dòng
    a, b, c, d = map(int, input().split())
    
    # Khởi tạo đối tượng với các hệ số tương ứng
    poly_sum = PolynomialSum(a, b, c, d)
    
    # In ra chuỗi biểu diễn đa thức tổng
    print(poly_sum.sum_str())
```

Xây dựng lớp Product(name, price). 
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức 
final_price() để trả về giá sau giảm. 
Công thức: price * (100 - percent) / 100 
 
Input 
name price percent 
 
Output 
In ra giá cuối cùng với 2 chữ số thập phân.

```python
import sys

# 1. Định nghĩa lớp cha Product
class Product:
    def __init__(self, name: str, price: float):
        self.name = name
        self.price = price

    def final_price(self) -> float:
        return self.price

# 2. Định nghĩa lớp con DiscountProduct kế thừa từ Product
class DiscountProduct(Product):
    def __init__(self, name: str, price: float, percent: float):
        # Gọi hàm khởi tạo của lớp cha để gán name và price
        super().__init__(name, price)
        self.percent = percent

    # Ghi đè (Override) phương thức final_price để tính giá sau giảm
    def final_price(self) -> float:
        return self.price * (100 - self.percent) / 100

# 3. Xử lý Input và Output
if __name__ == "__main__":
    # Đọc dữ liệu từ input
    input_data = sys.stdin.read().split()
    
    if len(input_data) >= 3:
        # Tách các thành phần dữ liệu (Tên, Giá gốc, Phần trăm giảm)
        name = input_data[0]
        price = float(input_data[1])
        percent = float(input_data[2])
        
        # Khởi tạo đối tượng từ lớp con
        product = DiscountProduct(name, price, percent)
        
        # In kết quả định dạng 2 chữ số thập phân
        print(f"{product.final_price():.2f}")
```

#### Bai 13
Yêu cầu 
Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), 
value(). 
Sau đó thực hiện dãy thao tác trên đối tượng. 
 
Input 
Dòng 1: n 
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách. 
 
Output 
Giá trị cuối cùng của count.

```python
import sys

# 1. Định nghĩa lớp Counter
class Counter:
    def __init__(self):
        # Thuộc tính count ban đầu bằng 0
        self.count = 0

    # Phương thức tăng giá trị
    def inc(self):
        self.count += 1

    # Phương thức giảm giá trị
    def dec(self):
        self.count -= 1

    # Phương thức trả về giá trị hiện tại
    def value(self) -> int:
        return self.count

# 2. Xử lý Input và Output
if __name__ == "__main__":
    # Đọc toàn bộ dữ liệu từ input
    input_data = sys.stdin.read().split()
    
    if input_data:
        # Dòng 1: Số lượng thao tác n
        n = int(input_data[0])
        
        # Dòng 2: Danh sách các thao tác (INC hoặc DEC)
        operations = input_data[1:n+1]
        
        # Khởi tạo đối tượng counter
        counter = Counter()
        
        # Duyệt qua từng thao tác và thực thi phương thức tương ứng
        for op in operations:
            if op == "INC":
                counter.inc()
            elif op == "DEC":
                counter.dec()
        
        # In ra giá trị cuối cùng bằng phương thức value()
        print(counter.value())
```

#### bai 14:
Yêu cầu 
Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây. 
 
Input 
h m s 
 
Output 
Tổng số giây. 

```python
import sys

# 1. Định nghĩa lớp Time
class Time:
    def __init__(self, h: int, m: int, s: int):
        self.h = h
        self.m = m
        self.s = s

    # Phương thức đổi thời gian sang tổng số giây
    def to_seconds(self) -> int:
        # 1 giờ = 3600 giây, 1 phút = 60 giây
        return self.h * 3600 + self.m * 60 + self.s

# 2. Xử lý Input và Output
if __name__ == "__main__":
    # Đọc dữ liệu từ input
    input_data = sys.stdin.read().split()
    
    if len(input_data) >= 3:
        # Ép kiểu dữ liệu sang số nguyên (int)
        h = int(input_data[0])
        m = int(input_data[1])
        s = int(input_data[2])
        
        # Khởi tạo đối tượng Time
        t = Time(h, m, s)
        
        # In ra tổng số giây kết quả
        print(t.to_seconds())
```

Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng. 

```python
import sys

# 1. Định nghĩa lớp Stack sử dụng list để lưu trữ phần tử
class Stack:
    def __init__(self):
        # Khởi tạo một danh sách rỗng để làm bộ lưu trữ cho Stack
        self.items = []

    # Thêm phần tử x vào đỉnh ngăn xếp
    def push(self, x):
        self.items.append(x)

    # Loại bỏ phần tử ở đỉnh ngăn xếp
    def pop(self):
        if not self.is_empty():
            return self.items.pop()

    # Xem giá trị của phần tử ở đỉnh ngăn xếp mà không loại bỏ nó
    def top(self):
        if not self.is_empty():
            return self.items[-1]

    # Kiểm tra ngăn xếp có rỗng hay không
    def is_empty(self) -> bool:
        return len(self.items) == 0

# 2. Xử lý Input và Output
if __name__ == "__main__":
    # Đọc tất cả các dòng từ input đầu vào
    input_lines = sys.stdin.read().splitlines()
    
    if input_lines:
        # Dòng đầu tiên chứa số lượng lệnh n
        n = int(input_lines[0].strip())
        
        # Khởi tạo đối tượng stack
        stack = Stack()
        
        # Duyệt qua từng dòng lệnh tiếp theo
        for i in range(1, n + 1):
            if i < len(input_lines):
                line = input_lines[i].strip()
                if not line:
                    continue
                
                # Tách lệnh thành các phần tử (ví dụ: ["PUSH", "5"] hoặc ["TOP"])
                parts = line.split()
                command = parts[0]
                
                if command == "PUSH":
                    x = parts[1]
                    stack.push(x)
                elif command == "POP":
                    stack.pop()
                elif command == "TOP":
                    # In ra phần tử đỉnh ngăn xếp theo yêu cầu output
                    print(stack.top())
```
