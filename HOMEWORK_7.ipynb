### Bai 1
Viết hàm tìm kiếm nhị phân 
trên một danh sách đã 
được sắp xếp tăng dần. 
Trả về chỉ số của phần tử 
hoặc -1 nếu không tìm 
thấy. mẫu hàm: def 
binary_search(arr, left, 
right, target):

```python
def binary_search(arr, left, right, target):
    if left > right:
        return -1
    mid = left + (right - left) // 2
    if arr[mid] == target:
        return mid

    elif arr[mid] > target:
        return binary_search(arr, left, mid - 1, target)

    else:
        return binary_search(arr, mid + 1, right, target)
    
arr = list(map(int,input().split()))
arr = sorted(arr)
print(binary_search(arr))
```

### Bai 2
Cho danh sách arr và một 
số nguyên x, đếm xem x 
xuất hiện bao nhiêu lần 
trong danh sách. def 
count_occurrences(arr, x) 
> int: 

```python
def count_occurrences(arr: list, x: int) -> int:
    count = 0
    for num in arr:
        if num == x:
            count += 1
    return count

arr = list(map(int,input().split()))
x = int(input())
print(count_occurrences(arr,x))
```

#### Bai 3
Viết hàm thực hiện thuật 
toán bubble sort, Selection 
Sort để sắp xếp danh sách 
theo thứ tự tăng dần. Sắp 
xếp nổi bọt (Bubble Sort); 

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = list(map(int,input().split()))
print(bubble_sort(arr))
```

#### Bai 4
Viết hàm tìm phần tử có số 
lần xuất hiện nhiều nhất 
trong danh sách. Nếu có 
nhiều phần tử cùng số lần, 
trả về phần tử xuất hiện 
trước. 

```python
from collections import Counter

def find_most_frequent(lst):
    if not lst:
        return None
    counts = Counter(lst)

    most_frequent = max(counts, key=counts.get)
    
    return most_frequent
arr = list(map(int,input().split()))
print(find_most_frequent(arr))
```

### Bai 5
Cho danh sách các số 
nguyên và một số X, đếm 
tất cả các cặp (a, b) sao 
cho a + b = X. c

```python
def count_pairs_brute_force(arr, X):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == X:
                count += 1
    return count
arr = list(map(int,input().split()))
X = int(input())
print(count_pairs_brute_force(arr, X))
```

### Bai 7
Cho một danh sách các số 
nguyên nums, hãy tìm độ 
dài của dãy con tăng 
nghiêm ngặt dài nhất 
(không nhất thiết phải liền 
kề).

```python
def length_of_LIS_dp(nums):
    if not nums:
        return 0
    dp = [1] * len(nums)
    
    for i in range(len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
                
    return max(dp)

# Thử nghiệm
nums = list(map(int,input().split()))
print(length_of_LIS_dp(nums))  
```

#### Bai 8
Bạn được cung cấp: một 
danh sách các khoảng thời 
gian interval = [[l1, r1], [l2, 
r2], ...] và một danh sách 
các truy vấn queries = [q1, 
q2, ...]  
Với mỗi truy vấn q, hãy tìm 
độ dài nhỏ nhất của 
khoảng thời gian [li, ri] sao 
cho li ≤ q ≤ ri.  
Nếu không tồn tại khoảng 
thời gian nào như vậy, hãy 
trả về -1 cho truy vấn đó. 
solve in < O(m.n), hint: 
binary search in the 
queries lis

```python
import heapq

def minInterval(intervals, queries):

    intervals.sort(key=lambda x: x[0])
    sorted_queries = sorted([(q, i) for i, q in enumerate(queries)])
    
    res = [-1] * len(queries)
    min_heap = [] 
    i = 0         
    n = len(intervals)
    for q, original_idx in sorted_queries:
        while i < n and intervals[i][0] <= q:
            l, r = intervals[i]
            length = r - l + 1
            heapq.heappush(min_heap, (length, r))
            i += 1
        while min_heap and min_heap[0][1] < q:
            heapq.heappop(min_heap)
        if min_heap:
            res[original_idx] = min_heap[0][0]
            
    return res

intervals = eval(input())
queries = list(map(int,input().split()))
print("Kết quả:", minInterval(intervals, queries))
```

#### Bai 9
Cho một danh sách các số 
nums, với mỗi phần tử 
nums[i], hãy đếm xem có 
bao nhiêu phần tử bên phải 
nhỏ hơn nums[i].  
Trả về danh sách các số 
đếm. 
Efficient solution: using 
heap or binary search tree 

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.count = 1 
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, val):
        if not self.root:
            self.root = Node(val)
            return 0
        
        current = self.root
        smaller_count = 0
        while True:
            if val <= current.val:
                current.count += 1
                if not current.left:
                    current.left = Node(val)
                    break
                current = current.left
            else:
                smaller_count += current.count
                if not current.right:
                    current.right = Node(val)
                    break
                current = current.right
        return smaller_count
def countSmaller(nums):
    if not nums:
        return []
        
    bst = BinarySearchTree()
    result = [0] * len(nums)
    for i in range(len(nums) - 1, -1, -1):
        result[i] = bst.insert(nums[i])
        
    return result

nums = list(map(int,input().split()))
print("Mảng đầu vào: ", nums)
print("Kết quả số đếm:", countSmaller(nums)) 
```

#### Bai 10
Cho một danh sách các số 
nguyên nums và một số 
nguyên m, hãy tìm tổng 
mảng con lớn nhất theo 
modulo m.  
Tức là, với bất kỳ mảng 
con nums[i:j] nào, hãy tính 
sum(nums[i:j]) % m và trả 
về kết quả lớn nhất có thể. 
Gợi ý: 
Duy trì tổng tiền tố (prefix 
sum) mod m và sử dụng 
tìm kiếm nhị phân để tìm 
tổng tiền tố nhỏ nhất chỉ 
lớn hơn tổng hiện tại một 
chút (để tối đa hóa hiệu số 
bao quanh).

```python
from bisect import insort, bisect_right

def max_subarray_sum_modulo(nums, m):
    prefix_sums = []
    
    max_mod_sum = 0
    current_sum = 0
    
    for num in nums:
        current_sum = (current_sum + num) % m
        max_mod_sum = max(max_mod_sum, current_sum)
        idx = bisect_right(prefix_sums, current_sum)
        
        if idx < len(prefix_sums):
            max_mod_sum = max(max_mod_sum, (current_sum - prefix_sums[idx] + m) % m)
        insort(prefix_sums, current_sum)
    return max_mod_sum

nums =list(map(int,input().split()))
m = int(input())
print("Tổng mảng con lớn nhất theo mod M là:", max_subarray_sum_modulo(nums, m))
```

#### Bai 13
ãy hài hòa là một dãy 
thỏa mãn điều kiện giá trị 
lớn nhất và giá trị nhỏ nhất 
của dãy khác nhau đúng 
bằng \( 1 \). 
Viết hàm trả về độ dài của 
dãy hài hòa con dài nhất có 
thể có trong mảng giá trị 
nguyên \( arr \) có \( n \) 
phần tử 

```python
def find_lhs(arr):
    counts = {}
    for num in arr:
        counts[num] = counts.get(num, 0) + 1
    max_length = 0
    for num in counts:
        if num + 1 in counts:
            current_length = counts[num] + counts[num + 1]
            max_length = max(max_length, current_length)
            
    return max_length



arr = list(map(int,input().split()))
print("Độ dài dãy hài hòa con dài nhất là:", find_lhs(arr))
```

```python
import sys

def find_number_seven():
    input_data = sys.stdin.read().splitlines()
    if len(input_data) < 2:
        return
    n = int(input_data[0])
    nums = list(map(int, input_data[1].split()))
    positions = []
    for i in range(n - 1, -1, -1):
        if nums[i] == 7:
            positions.append(str(i))
    if len(positions) == 0:
        print("Not found")
    else:
        print(" ".join(positions))

if __name__ == "__main__":
    find_number_seven()
```

Đầu ra: Tên của hai người 
ngồi cạnh Nemo theo định 
dạng "<name1> and 
<name2>" (name1 là tên 
của người đứng trước và 
name2 là tên của người 
đứng sau Nemo trong danh 
sách). 

```python
import sys

def find_nemo_neighbors():
    input_data = sys.stdin.read().splitlines()
    if not input_data:
        return
    n = int(input_data[0].strip())
    guests = []
    for i in range(1, n + 1):
        guests.append(input_data[i].strip())
    nemo_index = guests.index("Nemo")
    left_index = (nemo_index - 1 + n) % n
    right_index = (nemo_index + 1) % n
    name1 = guests[left_index]
    name2 = guests[right_index]
    print(f"{name1} and {name2}")

if __name__ == "__main__":
    find_nemo_neighbors()
```
