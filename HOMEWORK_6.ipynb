#### Bai 3
Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple 
sau khi xoay trái k vị trí. (Giữ nguyên thứ tự tương đối sau xoay; không sắp 
xếp.)

```python

def xoay_trai_tuple():
    input_tuple = tuple(map(int, input().split()))
    k = int(input())
    n = len(input_tuple)
    if n == 0 or k == 0:
        print(input_tuple)
        return 0
    k = k % n
    ket_qua = input_tuple[k:] + input_tuple[:k]
    print(ket_qua)
```

#### Bai 4
Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo 
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key 
lặp lại, thêm vào list của key đó 

```python
from collections import defaultdict

def tao_dict_nhom_value():
    chuoi_nhap = input().strip()
    if not chuoi_nhap:
        print({})
        return
    cac_cap = chuoi_nhap.split()
    ket_qua = defaultdict(list)
    
    for cap in cac_cap:
        key, value = cap.split(':', 1)

        ket_qua[key].append(value)
    print(dict(ket_qua))

if __name__ == "__main__":
    tao_dict_nhom_value()
```

#### Bai 5
Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 
'zeros': ...} (đếm số >0, <0, =0) 

```python
def dem_phan_loai_so():
    chuoi_nhap = input().strip()
    if not chuoi_nhap:
        ket_qua = {'positives': 0, 'negatives': 0, 'zeros': 0}
        print(ket_qua)
        return
    danh_sach_so = list(map(int, chuoi_nhap.split()))
    ket_qua = {'positives': 0, 'negatives': 0, 'zeros': 0}

    for so in danh_sach_so:
        if so > 0:
            ket_qua['positives'] += 1
        elif so < 0:
            ket_qua['negatives'] += 1
        else:
            ket_qua['zeros'] += 1
    print(ket_qua)
if __name__ == "__main__":
    dem_phan_loai_so()
```

#### Bai 9
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một 
dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ 
được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới 
theo thứ tự từ điển tăng dần của key.  

```python
import json
def nhap_dictionary(ten_dict):
    print(f"\n--- Nhập {ten_dict} ---")
    print("Nhập định dạng JSON (Ví dụ: {\"a\": 10, \"b\": 20, \"c\": 5})")
    while True:
        try:
            chuoi_nhap = input("Nhập chuỗi JSON: ")
            du_lieu = json.loads(chuoi_nhap)
            if isinstance(du_lieu, dict):
                return {k: float(v) for k, v in du_lieu.items()}
            else:
                print("Vui lòng nhập đúng định dạng một dictionary {}!")
        except json.JSONDecodeError:
            print("Chuỗi nhập không hợp lệ hoặc sai định dạng JSON. Vui lòng thử lại.")
dict1 = nhap_dictionary("Dictionary thứ nhất")
dict2 = nhap_dictionary("Dictionary thứ hai")
dict_gop = {}
for key, value in dict1.items():
    dict_gop[key] = value
for key, value in dict2.items():
    if key in dict_gop:
        dict_gop[key] += value  
    else:
        dict_gop[key] = value  
dict_sap_xep = {key: dict_gop[key] for key in sorted(dict_gop.keys())}
print("\n" + "="*30)
print(f"Dict 1: {dict1}")
print(f"Dict 2: {dict2}")
print(f"Dict sau khi gộp và sắp xếp tăng dần theo key:")
print(dict_sap_xep)
```

#### Bai 10
Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên 
dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là 
một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh 
sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được 
sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple. 

```python
def tim_cap_so(danh_sach, k):
    cac_cap_thoa_man = set()
    n = len(danh_sach)
    for i in range(n):
        for j in range(i + 1, n):
            if danh_sach[i] + danh_sach[j] == k:
                cap_so = (danh_sach[i], danh_sach[j])
                cac_cap_thoa_man.add(cap_so)
    ket_qua = sorted(list(cac_cap_thoa_man))
    
    return ket_qua

if __name__ == "__main__":
    try:
  
        input_list = input("Nhập danh sách các số nguyên (cách nhau bởi dấu cách): ")
        danh_sach_so = [int(x) for x in input_list.split()]
        

        k = int(input("Nhập số nguyên dương k: "))
        if k <= 0:
            print("Lưu ý: k phải là số nguyên dương theo yêu cầu đề bài.")
        

        ket_qua_cuoi = tim_cap_so(danh_sach_so, k)
        print(f"List các cặp số có tổng bằng {k} là:")
        print(ket_qua_cuoi)
        
    except ValueError:
        print("Vui lòng nhập đúng định dạng số nguyên!")
```

#### Bai 12
[Tìm phần tử xuất hiện nhiều nhất] 
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong 
list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ 
nhất trong số đó. 

```python
from collections import Counter

def tim_phan_tu_xuat_hien_nhieu_nhat(danh_sach):
    if not danh_sach:
        return None
    dem_tan_suat = Counter(danh_sach)
    so_can_tim = min(dem_tan_suat.items(), key=lambda x: (-x[1], x[0]))[0]
    
    return so_can_tim
if __name__ == "__main__":
    try:
        input_data = input("Nhập danh sách các số nguyên (cách nhau bởi dấu cách): ")
        danh_sach_so = [int(x) for x in input_data.split()]
        if danh_sach_so:
            ket_qua = tim_phan_tu_xuat_hien_nhieu_nhat(danh_sach_so)
            print(f"Số xuất hiện nhiều nhất (và nhỏ nhất nếu trùng tần suất) là: {ket_qua}")
        else:
            print("Danh sách trống!")
            
    except ValueError:
        print("Vui lòng chỉ nhập các số nguyên hợp lệ!")
```

#### Bai 15
Lọc dictionary theo giá trị] 
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số 
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp 
key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary 
mới. 

```python
import ast
dict_input = ast.literal_eval(input("Nhập dictionary: "))
k = int(input("Nhập số nguyên k: "))
filtered_dict = {key: value for key, value in dict_input.items() if value > k}
print(filtered_dict)
```
