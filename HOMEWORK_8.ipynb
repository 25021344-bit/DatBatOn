#### Bai 3
Sắp xếp có điều kiện: Số chẵn đứng 
trước, số lẻ đứng sau, chẵn tăng dần, lẻ 
giảm dần

```python
def conditional_sort_v1(arr):
    evens = [x for x in arr if x % 2 == 0]
    odds = [x for x in arr if x % 2 != 0]
    evens.sort()         
    odds.sort(reverse=True) 
    return evens + odds
arr = list(map(int,input().split()))
result = conditional_sort_v1(arr)
print(result)
```

#### Bai 4
Merge 2 mảng đã sắp xếp (có thể tăng 
dần, có thể giảm dần) mà không dùng 
bất cứ thuật toán sắp xếp nào thành một 
mảng cũng có cùng thứ tự sắp xếp như 2 
mảng ban đầu 

```python
def merge_sorted_arrays(arr1, arr2):

    if not arr1: return arr2
    if not arr2: return arr1
    is_ascending = True
    if len(arr1) > 1:
        is_ascending = arr1[0] <= arr1[-1]
    elif len(arr2) > 1:
        is_ascending = arr2[0] <= arr2[-1]

    res = []
    i, j = 0, 0
    len1, len2 = len(arr1), len(arr2)
    
    # 2. Dùng 2 con trỏ để duyệt qua cả 2 mảng
    while i < len1 and j < len2:
        if is_ascending:
            condition = arr1[i] <= arr2[j]
        else:
            condition = arr1[i] >= arr2[j]
            
        if condition:
            res.append(arr1[i])
            i += 1
        else:
            res.append(arr2[j])
            j += 1
    while i < len1:
        res.append(arr1[i])
        i += 1
    while j < len2:
        res.append(arr2[j])
        j += 1
        
    return res
a_inc = list(map(int,input().split()))
b_inc = list(map(int,input().split()))
print(merge_sorted_arrays(a_inc, b_inc)) 
```

#### Bai 7
Cho N tên sinh viên gồm cả họ và tên, 
tìm xem tên nào hay gặp nhất. Nếu có 
nhiều tên cùng số lần gặp thì lấy tên có 
thứ tự từ điển nhỏ nhất 

```python
def find_most_common_name(student_list):
    name_counts = {}
    
    for full_name in student_list:

        parts = full_name.strip().split()
        if not parts:
            continue
    
        first_name = parts[-1]
        

        name_counts[first_name] = name_counts.get(first_name, 0) + 1

    best_name = ""
    max_count = -1
    

    for name, count in name_counts.items():
        if count > max_count:
            max_count = count
            best_name = name
        elif count == max_count:
            if name < best_name:
                best_name = name
                
    return best_name, max_count

k = int(input())
n = []
while k>0:
    n.append(str(input()))
    if len(n) == k:
        break

name, count = find_most_common_name(k)
print(f"Tên hay gặp nhất: {name} (xuất hiện {count} lần)")
```

#### Bai 8
Đếm số lượng tên trong xâu kí tự. Tên 
được định nghĩa là xâu có độ dài >=2 và 
có chữ cái đầu viết hoa, các chữ cái khác 
viết thường, chỉ có kí tự cái.

```python
def count_valid_names(s):

    words = s.split()
    count = 0
    
    for word in words:

        if len(word) < 2:
            continue
        if not word[0].isupper():
            continue
        if all(char.islower() for char in word[1:]):
            count += 1
            
    return count


S = str(input())
print("Số lượng tên hợp lệ:", count_valid_names(S))
```

### Bai 9
Cho xâu kí tự, xóa các từ giống nhau liền 
kề, chỉ giữ lại từ đầu tiên 

```python
def remove_adjacent_duplicates(s):

    words = s.split()
    
    if not words:
        return ""
    res = [words[0]]

    for i in range(1, len(words)):

        if words[i] != res[-1]:
            res.append(words[i])

    return " ".join(res)


S = str(input())
print(remove_adjacent_duplicates(S))
```

### Bai 10
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra 
xem dãy ngoặc đã cho có hợp lệ hay 
không 

```python
def is_valid_parentheses(s):

    stack = []

    mapping = {")": "(", "}": "{", "]": "["}
    
    for char in s:

        if char in mapping:

            top_element = stack.pop() if stack else '#'

            if mapping[char] != top_element:
                return False
        else:

            stack.append(char)
            
    return len(stack) == 0

test_cases = list(map(str,input().split()))

for test in test_cases:
    print(f"Chuỗi '{test}': {is_valid_parentheses(test)}")
```

#### bai 11
Bà ̀ i toán Tìm dãy các hợp số dài nhất liên 
tiếp không chứa số nguyên nào <= N (N 
là số nguyên dương cho trước). Nếu có 
nhiều dãy cùng độ dài thì lấy dãy nhỏ 
nhất (gặp đầu tiên từ trái qua) 

```python
def find_longest_composite_sequence(n):
    if n < 4:
        return "Không có dãy hợp số nào hợp lệ."

    is_prime = [True] * (n + 1)
    is_prime[0] = is_prime[1] = False
    
    for i in range(2, int(n**0.5) + 1):
        if is_prime[i]:
            for j in range(i*i, n + 1, i):
                is_prime[j] = False

    primes = [i for i, prime in enumerate(is_prime) if prime]

    max_len = 0
    start_num = -1
    end_num = -1

    for i in range(len(primes) - 1):
        p1 = primes[i]
        p2 = primes[i+1]

        current_len = p2 - p1 - 1

        if current_len > max_len:
            max_len = current_len
            start_num = p1 + 1
            end_num = p2 - 1

    if max_len == 0:
        return "Không tìm thấy dãy hợp số."
        
    return {
        "Do_dai": max_len,
        "Tu_so": start_num,
        "Den_so": end_num,
        "Day_so": list(range(start_num, end_num + 1))
    }

N = int(input())
result = find_longest_composite_sequence(N)
print(f"Với N = {N}:")
print(f"- Độ dài dài nhất: {result['Do_dai']}")
print(f"- Dãy số: {result['Day_so']} (nằm giữa số nguyên tố {result['Tu_so']-1} và {result['Den_so']+1})")
```

của một danh sách nhập vào từ bàn 
phím. Lấy dãy đầu tiên nếu có nhiều dãy 
có số phần tử bằng nhau 

```python
def find_longest_increasing_continuous_subsequence():
    input_str = input("Nhập các số trong danh sách, cách nhau bởi dấu cách: ")

    A = [int(x) for x in input_str.split()]

    if not A:
        print("Danh sách rỗng!")
        return

    max_len = 1       
    best_start = 0    
    
    curr_len = 1      
    curr_start = 0    

    for i in range(1, len(A)):
        if A[i] > A[i-1]:

            curr_len += 1
        else:

            if curr_len > max_len:
                max_len = curr_len
                best_start = curr_start
            

            curr_len = 1
            curr_start = i


    if curr_len > max_len:
        max_len = curr_len
        best_start = curr_start


    longest_subsequence = A[best_start : best_start + max_len]


    print("\n--- KẾT QUẢ ---")
    print(f"Danh sách đã nhập: {A}")
    print(f"Độ dài dãy con tăng liên tục dài nhất: {max_len}")
    print(f"Dãy con đầu tiên tìm được: {longest_subsequence}")


find_longest_increasing_continuous_subsequence()
```

#### Bai 13
Tìm Dãy con tăng dần KHÔNG liên tục 
dài nhất của một danh sách nhập vào từ 
bàn phím  

```python
def find_lis(A):
    n = len(A)
    if n == 0:
        return []

    L = [1] * n

    T = [-1] * n
    for i in range(1, n):
        for j in range(i):
            if A[j] < A[i] and L[i] < L[j] + 1:
                L[i] = L[j] + 1
                T[i] = j

    max_length = max(L)
    max_index = L.index(max_length)


    lis_subsequence = []
    curr = max_index
    while curr != -1:
        lis_subsequence.append(A[curr])
        curr = T[curr]

    lis_subsequence.reverse()

    return lis_subsequence



if __name__ == "__main__":
    print("=== TÌM DÃY CON TĂNG DÀI NHẤT (KHÔNG LIÊN TỤC) ===")
    try:
       
        input_str = input("Nhập các số trong danh sách (cách nhau bằng dấu cách): ")

   
        A = [int(x) for x in input_str.split()]

        if not A:
            print("Danh sách trống!")
        else:
            result = find_lis(A)
            print("-" * 40)
            print(f"Mảng ban đầu: {A}")
            print(f"Độ dài dãy con tăng dài nhất: {len(result)}")
            print(f"Dãy con tìm được là: {result}")

    except ValueError:
        print("Vui lòng chỉ nhập các số nguyên hợp lệ!")
```

#### bai 14
Đổi chỗ đúng 2 chữ số để được số nhỏ 
nhất có thể

```python
def doi_cho_nho_nhat(n_str):

    s = list(n_str)
    n = len(s)

    min_num = n_str

    for i in range(n):
        for j in range(i + 1, n):
            s[i], s[j] = s[j], s[i]
            if s[0] != '0':
                current_num = "".join(s)
                if current_num < min_num:
                    min_num = current_num
            s[i], s[j] = s[j], s[i]
    return min_num



a = input()
print(f"{doi_cho_nho_nhat(a)}")
```
